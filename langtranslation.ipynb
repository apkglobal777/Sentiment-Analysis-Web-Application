```python
#Language Translation Using using Hugging face
from transformers import MarianMTModel,MarianTokenizer
#Select the model for translation from huggin face
model_name='Helsinki-NLP/opus-mt-en-hi'
#Load the model using model name
model=MarianMTModel.from_pretrained(model_name)
#Load the tokenizer using model name
tokenizer=MarianTokenizer.from_pretrained(model_name)
#Get the input promp text from uer to translate
promt=input("Enter text to translate")
#encode the input text into input tokens using tokenizer e.g Hi->0 1
input_tokens=tokenizer(promt, return_tensors='pt',padding=True)
#Generate the output tokens using model via data process e.g 40-> Kaise
output_tokens=model.generate(**input_tokens)
#Decode the ouput token suing tokenizer and generate the final tanslated text
result=tokenizer.decode(output_tokens[0], skip_special_tokens=True)

#Print the result
print(result)
```
